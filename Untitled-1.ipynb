```python
df['appliance_type'].unique()
```

```python
df['appliance_type']=df['appliance_type'].str.strip()
df['appliance_type']=df['appliance_type'].str.title()
df["appliance_type"]=df['appliance_type'].replace({'AC':'Air Conditioner','refrigerator':'Refrigerator','washing machine':'Washing Machine','air conditioner':'Air Conditioner'})
df["appliance_type"]=df['appliance_type'].fillna(df['appliance_type'].mode()[0])
```

```python
df['appliance_type'].unique()
```

```python
df['usage_hours'].unique()
```

```python
df['usage_hours']=df['usage_hours'].fillna(df['usage_hours'].median())
```

```python
df['room_size_m2'].unique()
```

```python
df['room_size_m2']=df['room_size_m2'].replace({'?':None,'missing':None,"unknown":None})
df['room_size_m2']=pd.to_numeric(df['room_size_m2'],errors="coerce")
df['room_size_m2']=df['room_size_m2'].fillna(df['room_size_m2'].median())
df['room_size_m2']=df['room_size_m2'].astype(int)
```

```python
df['appliance_power_watts'].unique()
```

```python
df['appliance_power_watts'] = pd.to_numeric(
    df['appliance_power_watts'].astype(str).str.replace('W', '', regex=False).str.strip(), 
    errors='coerce')
df.loc[(df['appliance_power_watts'])<=0 | (df['appliance_power_watts']>4000),'appliance_power_watts']=np.nan
df['appliance_power_watts']=df['appliance_power_watts'].fillna(df['appliance_power_watts'].median()).astype(int)
```

```python
df['usage_hours'].unique()
```

```python
df.loc[(df['usage_hours'])<=0 | (df['usage_hours']>1.0),'usage_hours']=np.nan
df['usage_hours']=df['usage_hours'].fillna(df['usage_hours'].median())
```

```python
df['temperature_c'].unique()
```

```python
df.loc[(df['temperature_c'])<10 | (df['temperature_c']>50),'temperature_c']=np.nan
df['temperature_c']=df['temperature_c'].fillna(df["temperature_c"].mean())
```

```python
df['number_of_people'].unique()
```

```python

df['number_of_people']=df['number_of_people'].replace({'missing':None,'unknown':None,'?':None})
df['number_of_people']=df['number_of_people'].fillna(df['number_of_people'].mode()[0])
df['number_of_people']=df['number_of_people'].astype(int)
```

```python
df['time_of_day'].unique()
```

```python
df['time_of_day']=df['time_of_day'].astype(str).str.lower().str.strip()
df['time_of_day']=df['time_of_day'].replace({'Afternoon': 'afternoon',' EVENING ': 'evening','Evening': 'evening','morning': 'morning', 'Morning': 'morning','Night': 'night'})
df['time_of_day']=df['time_of_day'].fillna(df['time_of_day'].mode()[0])
```

```python
df['energy_consumption_kwh'].unique()
```

```python
df['energy_consumption_kwh'] = df['energy_consumption_kwh'].replace(['?', 'N/A', 'unknown', 'missing', ''], np.nan)
df['energy_consumption_kwh']=df['energy_consumption_kwh'].astype(float)
df.loc[(df['energy_consumption_kwh']<0.1 )| (df['energy_consumption_kwh']>25.0),'energy_consumption_kwh']=np.nan
df['energy_consumption_kwh']=df['energy_consumption_kwh'].fillna(df['energy_consumption_kwh'].median())
```

```python
df['previous_consumption_kwh'].unique()
```

```python

df['previous_consumption_kwh'] = df['previous_consumption_kwh'].replace({'?': None,"missing":None,"unknown":None})
df['previous_consumption_kwh'] = df['previous_consumption_kwh'].astype(float)
df.loc[(df['previous_consumption_kwh'] < 0.1) | (df['previous_consumption_kwh'] > 25.0), 'previous_consumption_kwh'] = np.nan
df['previous_consumption_kwh'] = df['previous_consumption_kwh'].fillna(df['previous_consumption_kwh'].median())
```

train model

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score,accuracy_score,root_mean_squared_error
from sklearn.preprocessing import StandardScaler
from sklearn.tree import  DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor,GradientBoostingRegressor
from xgboost import XGBRegressor
from sklearn.model_selection import GridSearchCV
```

```python
x=df_encoding.drop(columns=['energy_consumption_kwh'],errors='ignore')
y=df_encoding['energy_consumption_kwh']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
xgb=XGBRegressor(max_depth=3,random_state=42,learning_rate=0.05,n_estimators=100)
xgb.fit(x_train,y_train)
```

```python
y_pred_test=xgb.predict(x_test)
y_pred_train=xgb.predict(x_train)
r2_train=r2_score(y_train,y_pred_train)
r2_test=r2_score(y_test,y_pred_test)
mae_train=mean_absolute_error(y_train,y_pred_train)
mae_test=mean_absolute_error(y_test,y_pred_test)
rmse_train=root_mean_squared_error(y_train,y_pred_train)
rmse_test=root_mean_squared_error(y_test,y_pred_test)

print('r2_train',r2_train)
print('r2_test',r2_test)
print('MAE_train',mae_train)
print('MAE_test',mae_test)

print('RMSE_train',rmse_train)
print('RMSE_test',rmse_test)
```

```python
xgb_r2_tr = r2_train
xgb_r2_te = r2_test
xgb_rmse_te = rmse_test
xgb_mae_te = mae_test
```

```python
dt=DecisionTreeRegressor(max_depth=3,random_state=42)
dt.fit(x_train,y_train)
```

```python
y_pred_test=dt.predict(x_test)
y_pred_train=dt.predict(x_train)
r2_train=r2_score(y_train,y_pred_train)
r2_test=r2_score(y_test,y_pred_test)
mae_train=mean_absolute_error(y_train,y_pred_train)
mae_test=mean_absolute_error(y_test,y_pred_test)
rmse_train=root_mean_squared_error(y_train,y_pred_train)
rmse_test=root_mean_squared_error(y_test,y_pred_test)

print('r2_train',r2_train)
print('r2_test',r2_test)
print('MAE_train',mae_train)
print('MAE_test',mae_test)

print('RMSE_train',rmse_train)
print('RMSE_test',rmse_test)
```

```python
dt_r2_tr = r2_train
dt_r2_te = r2_test
dt_rmse_te = rmse_test
dt_mae_te = mae_test
```

```python
rf=RandomForestRegressor(n_estimators=50,max_depth=3,random_state=42)
rf.fit(x_train,y_train)
```

```python
y_pred_test=rf.predict(x_test)
y_pred_train=rf.predict(x_train)
r2_train=r2_score(y_train,y_pred_train)
r2_test=r2_score(y_test,y_pred_test)
mae_train=mean_absolute_error(y_train,y_pred_train)
mae_test=mean_absolute_error(y_test,y_pred_test)
rmse_train=root_mean_squared_error(y_train,y_pred_train)
rmse_test=root_mean_squared_error(y_test,y_pred_test)

print('r2_train',r2_train)
print('r2_test',r2_test)
print('MAE_train',mae_train)
print('MAE_test',mae_test)

print('RMSE_train',rmse_train)
print('RMSE_test',rmse_test)
```

```python
rf_r2_tr = r2_train
rf_r2_te = r2_test
rf_rmse_te = rmse_test
rf_mae_te = mae_test
```

```python
gd=GradientBoostingRegressor(max_depth=3,random_state=42,n_estimators=100,learning_rate=0.05)
gd.fit(x_train,y_train)
```

```python
y_pred_test=gd.predict(x_test)
y_pred_train=gd.predict(x_train)
r2_train=r2_score(y_train,y_pred_train)
r2_test=r2_score(y_test,y_pred_test)
mae_train=mean_absolute_error(y_train,y_pred_train)
mae_test=mean_absolute_error(y_test,y_pred_test)
rmse_train=root_mean_squared_error(y_train,y_pred_train)
rmse_test=root_mean_squared_error(y_test,y_pred_test)

print('r2_train',r2_train)
print('r2_test',r2_test)
print('MAE_train',mae_train)
print('MAE_test',mae_test)

print('RMSE_train',rmse_train)
print('RMSE_test',rmse_test)
```

```python
gb_r2_tr = r2_train
gb_r2_te = r2_test
gb_rmse_te = rmse_test
gb_mae_te = mae_test
```

```python

results = pd.DataFrame({
    'Model': ['Decision Tree', 'Random Forest', 'Gradient Boosting', 'XGBoost'],
    'R2 Train': [dt_r2_tr, rf_r2_tr, gb_r2_tr, xgb_r2_tr],
    'R2 Test':  [dt_r2_te, rf_r2_te, gb_r2_te, xgb_r2_te],
    'RMSE Test': [dt_rmse_te, rf_rmse_te, gb_rmse_te, xgb_rmse_te],
    'MAE Test':  [dt_mae_te, rf_mae_te, gb_mae_te, xgb_mae_te]
})

print(results)
```

```python

best_model_name = results.loc[results['R2 Test'].idxmax(), 'Model']
print(f"Best Model: {best_model_name}")
```

HyperTuning

```python
param_grid={
'max_depth':[3,4,5],
'learning_rate':[0.01,0.05,0.1],
'n_estimators':[50,100,150]

}
grid=GridSearchCV(estimator=GradientBoostingRegressor(random_state=42),
   param_grid=param_grid,
   cv=5,
   n_jobs=-1,
   scoring='r2')
grid.fit(x_train,y_train)
print("Best Parameters:", grid.best_params_)
print("Best R2 Score after Tuning:", grid.best_score_)

best_gb = grid.best_estimator_
```
